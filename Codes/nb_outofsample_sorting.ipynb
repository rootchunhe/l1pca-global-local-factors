```python
from port_sort import load_signal_sorting_data, run_signal_sorting_pack, print_latex_signal_sorting_table

# Signal selection criteria:
# 1. Exclude accounting-based variables (slow-moving Compustat fundamentals)
# 2. Exclude signals with lookback horizon > 1 year
#    except betabab_1260d and seas_1_1na
# 3. Exclude microcap liquidity proxies (zero-trade measures)

low_risk_chars = [
    "beta_dimson_21d",
    "betabab_1260d",
    "betadown_252d",
    "ivol_capm_21d",
    "ivol_capm_252d",
    "ivol_ff3_21d",
    "ivol_hxz4_21d",
    "rmax1_21d",
    "rmax5_21d",
    "rvol_21d",
    "turnover_126d",
]

seasonality_chars = [
    "coskew_21d",
]

short_reversal_chars = [
    "iskew_capm_21d",
    "iskew_ff3_21d",
    "iskew_hxz4_21d",
    "ret_1_0",
    "rmax5_rvol_21d",
    "rskew_21d",
]

momentum_chars = [
    "prc_highprc_252d",
    "resff3_6_1",
    "resff3_12_1",
    "ret_3_1",
    "ret_6_1",
    "ret_9_1",
    "ret_12_1",
    "seas_1_1na",
]

# Known direction by signal family:
# - bottom decile for low-risk / seasonality / short-term reversal
# - top decile for momentum
long_bottom_decile = low_risk_chars + seasonality_chars + short_reversal_chars
long_top_decile = momentum_chars
```

```python
data = load_signal_sorting_data(lnf_col="lnf_exposure_proxy")
detail, summary, meta = run_signal_sorting_pack(
    data=data,
    long_bottom_decile=long_bottom_decile,
    long_top_decile=long_top_decile,
)
```

```python
print_latex_signal_sorting_table(
    summary_df=summary,
    chars=["lnf_exposure_proxy"] + low_risk_chars,
    side="bottom",
    latex_filename="sorting_low_risk_bottom.tex",
)
```
